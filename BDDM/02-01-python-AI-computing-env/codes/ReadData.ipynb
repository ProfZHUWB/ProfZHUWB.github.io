2. 如果你[天池notebook](https://tianchi.aliyun.com/notebook-ai?lang=zh-cn) 上运行本 notebook 尝试手动下载数据并上传到保存本 notebook 的目录中

```python
base_url = 'https://profzhuwb.gitee.io/BDDM'
data_url = base_url + '/02-01-python-AI-computing-env/codes/02-01-data.zip_'
filename_encoding = 'gbk'
data_dir = '02-01-data'

def get_from_url(url):
    # 从 URL 提取文件名字
    # Extract filename from URL
    file_name = url.split('/')[-1]

    # 如果本地有同名文件，不下载
    # Do not download if file with same name exists locally
    import os
    if os.path.exists(file_name):
        print(f'文件：{file_name} 已存在，跳过下载')
        return file_name

    import urllib.request
    from shutil import copyfileobj
    import time
    # OpenerDirector 会自动检测系统的 Proxy 并使用它
    # OpenerDirector will automatically detect proxy and use it
    opener = urllib.request.build_opener()
    # 有些网站如 gitee.com 需要设置 User-Agent，否则拒绝访问
    # Sites such as gitee.com require User-Agent header, or reject access
    opener.addheaders = [('User-Agent', 'Mozilla/5.0')] # HTTPError: HTTP Error 403: Forbidden
    req = urllib.request.Request(url)
    print(f'下载自 {url}')
    start_time = time.time()
    with opener.open(req) as response, open(file_name, 'wb') as file:
        copyfileobj(response, file)
        end_time = time.time()
        print(f'保存到：{file_name}  ({end_time-start_time:.2f} s)')
    return file_name

# zip_file.extractall(data_dir) may fail to extract correct filename containing
# Chinese characters, when zip file is created on Chinese version of Windows
# Create a unzip function to correctly handle filenames with Chinese encoding
# 在中文版 Windows 上创建的文件名中的中文用的是 GBK，但是 Python 会用 cp437 解码导致产生乱码
# 自定义一个 unzip 方法，正确解码文件名字中的中文字符
def unzip(zip_filename, unzip_dir, encoding='gbk'):
    import zipfile
    with zipfile.ZipFile(zip_filename, 'r') as zip_file:
        import os
        os.makedirs(data_dir, exist_ok=True)
        for info in zip_file.filelist:
            filename = info.filename
            if info.flag_bits & 0x800 == 0:
                new_filename = filename.encode('cp437').decode(encoding)
            else:
                new_filename = filename
            fp = zip_file.open(filename)
            from shutil import copyfileobj
            dest_filename = data_dir+"/"+new_filename
            with open(dest_filename, 'wb') as file:
                copyfileobj(fp, file)
                print(f'解压 {dest_filename}')

unzip(get_from_url(url=data_url), data_dir, encoding=filename_encoding)
```
